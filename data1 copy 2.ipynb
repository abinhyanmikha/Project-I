## 3. Data cleaning and validation
Use this order:
1. Inspect the raw data: check columns, types, missing values, and duplicates.
2. Remove exact duplicate rows.
3. Fix data types: convert year, scores, counts, and sales to numeric. Values like "tbd" in User_Score will become missing.
4. Clean invalid values: check for impossible score ranges, negative sales, suspicious release years, and inconsistent category text. Correct values when you can; otherwise mark them missing or document why you remove a row.
5. Remove rows missing Global_Sales: it’s the value you want to predict, so rows without it can’t train the model.
6. Handle rows missing Name: remove them if you need game names for your analysis. Name is not a predictor in the current notebook, so this choice is about data quality and reporting.
7. Keep missing predictor values for now. Don’t delete those rows automatically.
8. Split into training and test sets.
9. Fill missing predictors using training data only: numeric values with training-set medians, and missing categories with "Unknown". Apply the same fill values to the test set.
10. Check the cleaned data: confirm duplicates and invalid values are handled, the target has no missing values, and predictors are ready for modeling.
So, cleaning starts with inspection, duplicates, types, and invalid values. Filling predictor gaps is a later preprocessing step, after the split.

```python
from sklearn.model_selection import train_test_split

# Features available in the selected post-review prediction scenario.
numeric_features = [
    "Game_Age",
    "Has_Review_Scores",
    "Review_Total_Count",
    "Critic_User_Gap",
    "Critic_Score",
    "Critic_Count",
    "User_Score",
    "User_Count",
    "Is_Major_Publisher"
]

categorical_features = [
    "Platform",
    "Genre",
    "Rating",
    "Publisher",
    "Developer"
]

features = numeric_features + categorical_features
target = "Global_Sales"

X = df[features].copy()
y = df[target].copy()

# Random holdout for a comparable baseline benchmark.
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
print("Training features:", X_train.shape[1])
```

```python

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# ==================================================
# 1. SEPARATE NUMERICAL AND CATEGORICAL COLUMNS
# ==================================================

X_train_num = X_train[numeric_features].copy()
X_test_num = X_test[numeric_features].copy()

X_train_cat = X_train[categorical_features].copy()
X_test_cat = X_test[categorical_features].copy()

# Check grouping columns
assert X_train[["Genre", "Game_Age"]].notna().all().all()
assert X_test[["Genre", "Game_Age"]].notna().all().all()


# ==================================================
# 2. NUMERICAL GROUP-WISE IMPUTATION
#    Group: Genre + Game_Age
#    Strategy: Median
# ==================================================

numeric_group_values = {}
numeric_global_values = {}

for col in numeric_features:

    # Global median fitted on training data
    numeric_imputer = SimpleImputer(strategy="median")
    numeric_imputer.fit(X_train_num[[col]])

    global_value = numeric_imputer.statistics_[0]

    if pd.isna(global_value):
        raise ValueError(
            f"{col} contains no valid training values."
        )

    numeric_global_values[col] = global_value

    # Create grouping data
    group_data = X_train[["Genre", "Game_Age"]].copy()
    group_data[col] = X_train_num[col]

    # Calculate group-wise median
    for group, group_df in group_data.groupby(
        ["Genre", "Game_Age"]
    ):

        valid_data = group_df[[col]].dropna()

        # Skip groups where all values are missing
        if valid_data.empty:
            continue

        group_imputer = SimpleImputer(strategy="median")
        group_imputer.fit(valid_data)

        numeric_group_values[(col, group)] = (
            group_imputer.statistics_[0]
        )


def impute_numeric(df, reference_df):

    df = df.copy()

    for col in numeric_features:

        missing_indices = df.index[df[col].isna()]

        for idx in missing_indices:

            group = (
                reference_df.loc[idx, "Genre"],
                reference_df.loc[idx, "Game_Age"]
            )

            # Group median or global median fallback
            value = numeric_group_values.get(
                (col, group),
                numeric_global_values[col]
            )

            df.loc[idx, col] = value

    return df


# Apply numerical imputation
X_train_num = impute_numeric(X_train_num, X_train)
X_test_num = impute_numeric(X_test_num, X_test)


# ==================================================
# 3. CATEGORICAL GROUP-WISE IMPUTATION
#    Group: Genre + Game_Age
#    Strategy: Most Frequent
# ==================================================

categorical_group_values = {}
categorical_global_values = {}

# Do not impute Genre because it is a grouping column
columns_to_impute = [
    col for col in categorical_features
    if col != "Genre"
]

for col in columns_to_impute:

    # Global most frequent value
    categorical_imputer = SimpleImputer(
        strategy="most_frequent"
    )

    categorical_imputer.fit(X_train_cat[[col]])

    global_value = categorical_imputer.statistics_[0]

    if pd.isna(global_value):
        raise ValueError(
            f"{col} contains no valid training values."
        )

    categorical_global_values[col] = global_value

    # Create grouping data
    group_data = X_train[["Genre", "Game_Age"]].copy()
    group_data[col] = X_train_cat[col]

    # Calculate group-wise most frequent value
    for group, group_df in group_data.groupby(
        ["Genre", "Game_Age"]
    ):

        valid_data = group_df[[col]].dropna()

        # Skip groups where all values are missing
        if valid_data.empty:
            continue

        group_imputer = SimpleImputer(
            strategy="most_frequent"
        )

        group_imputer.fit(valid_data)

        categorical_group_values[(col, group)] = (
            group_imputer.statistics_[0]
        )


def impute_categorical(df, reference_df):

    df = df.copy()

    for col in columns_to_impute:

        missing_indices = df.index[df[col].isna()]

        for idx in missing_indices:

            group = (
                reference_df.loc[idx, "Genre"],
                reference_df.loc[idx, "Game_Age"]
            )

            # Group mode or global mode fallback
            value = categorical_group_values.get(
                (col, group),
                categorical_global_values[col]
            )

            df.loc[idx, col] = value

    return df


# Apply categorical imputation
X_train_cat = impute_categorical(X_train_cat, X_train)
X_test_cat = impute_categorical(X_test_cat, X_test)


# ==================================================
# 4. ONE-HOT ENCODING
# ==================================================

encoder = OneHotEncoder(
    handle_unknown="ignore",
    min_frequency=10,
    sparse_output=False
)

X_train_encoded = encoder.fit_transform(X_train_cat)
X_test_encoded = encoder.transform(X_test_cat)

encoded_columns = encoder.get_feature_names_out(
    categorical_features
)


# ==================================================
# 5. COMBINE NUMERICAL AND CATEGORICAL FEATURES
# ==================================================

X_train_processed = pd.concat(
    [
        X_train_num,
        pd.DataFrame(
            X_train_encoded,
            columns=encoded_columns,
            index=X_train.index
        )
    ],
    axis=1
)

X_test_processed = pd.concat(
    [
        X_test_num,
        pd.DataFrame(
            X_test_encoded,
            columns=encoded_columns,
            index=X_test.index
        )
    ],
    axis=1
)


# ==================================================
# 6. LOG-TRANSFORM TARGET
# ==================================================

y_train_log = np.log1p(y_train)
y_test_log = np.log1p(y_test)


# ==================================================
# 7. CHECK RESULTS
# ==================================================

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)

print("\nRemaining missing values:")
print("Train:", X_train_processed.isna().sum().sum())
print("Test:", X_test_processed.isna().sum().sum())

print("\nNumerical missing values:")
print(X_train_num.isna().sum()[X_train_num.isna().sum() > 0])

print("\nCategorical missing values:")
print(X_train_cat.isna().sum()[X_train_cat.isna().sum() > 0])

assert X_train_processed.notna().all().all()
assert X_test_processed.notna().all().all()

print("\nPreprocessing completed successfully!")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from xgboost import XGBRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42),
    "XGBoost": XGBRegressor(random_state=42)
}

trained_models = {}

for name, model in models.items():
    model.fit(X_train_processed, y_train_log)
    trained_models[name] = model
```

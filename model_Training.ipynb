```python
from sklearn.model_selection import train_test_split

# Features available in the selected post-review prediction scenario.
numeric_features = [
    "Game_Age",
    "Has_Review_Scores",
    "Review_Total_Count",
    "Critic_User_Gap",
    "Critic_Score",
    "Critic_Count",
    "User_Score",
    "User_Count",
    "Is_Major_Publisher"
]

categorical_features = [
    "Platform",
    "Genre",
    "Rating",
    "Publisher",
    "Developer"
]

features = numeric_features + categorical_features
target = "Global_Sales"


X = df[features].copy()
y = df[target].copy()

# Random holdout for a comparable baseline benchmark.
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
print("Training features:", X_train.shape[1])
```

```python
df.isnull().sum()
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
import numpy as np

# 1. Handle missing numerical values
num_imputer = SimpleImputer(strategy="median")

X_train_num = pd.DataFrame(
    num_imputer.fit_transform(X_train[numeric_features]),
    columns=numeric_features,
    index=X_train.index
)

X_test_num = pd.DataFrame(
    num_imputer.transform(X_test[numeric_features]),
    columns=numeric_features,
    index=X_test.index
)

# 2. Handle missing categorical values
cat_imputer = SimpleImputer(strategy="most_frequent")

X_train_cat = cat_imputer.fit_transform(X_train[categorical_features])
X_test_cat = cat_imputer.transform(X_test[categorical_features])

# 3. Encode categorical features
encoder = OneHotEncoder(
    handle_unknown="ignore",
    min_frequency=10,
    sparse_output=False
)

X_train_encoded = encoder.fit_transform(X_train_cat)
X_test_encoded = encoder.transform(X_test_cat)

# 4. Combine numerical and encoded categorical features
encoded_columns = encoder.get_feature_names_out(categorical_features)

X_train_processed = pd.concat([
    X_train_num,
    pd.DataFrame(X_train_encoded, columns=encoded_columns, index=X_train.index)
], axis=1)

X_test_processed = pd.concat([
    X_test_num,
    pd.DataFrame(X_test_encoded, columns=encoded_columns, index=X_test.index)
], axis=1)

# 5. Log-transform target
y_train_log = np.log1p(y_train)
y_test_log = np.log1p(y_test)

print("Training shape:", X_train_processed.shape)
print("Testing shape:", X_test_processed.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from xgboost import XGBRegressor

models = {
    "Linear Regression": LinearRegression(),
    "Decision Tree": DecisionTreeRegressor(random_state=42),
    "Random Forest": RandomForestRegressor(random_state=42),
    "XGBoost": XGBRegressor(random_state=42)
}

trained_models = {}

for name, model in models.items():
    model.fit(X_train_processed, y_train_log)
    trained_models[name] = model
```

```python
results = []

for name, model in trained_models.items():
    predictions_log = model.predict(X_test_processed)
    predictions = np.maximum(np.expm1(predictions_log), 0)

    results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_test, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_test, predictions)),
        "R2": r2_score(y_test, predictions)
    })

results_df = pd.DataFrame(results).sort_values("MAE")
display(results_df)

best_initial_name = results_df.iloc[0]["Model"]
print("Best initial model:", best_initial_name)
```

```python
# Select the best-performing model from the comparison table.
best_initial_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_initial_name]

print("Best model based on test MAE:", best_initial_name)
```

```python
best_predictions_log = best_model.predict(X_test_processed)
best_predictions = np.maximum(np.expm1(best_predictions_log), 0)

best_result = pd.DataFrame([{
    "Model": best_initial_name,
    "MAE": mean_absolute_error(y_test, best_predictions),
    "RMSE": np.sqrt(mean_squared_error(y_test, best_predictions)),
    "R2": r2_score(y_test, best_predictions)
}])

display(best_result)
```
